# Eligible product counts — OCI Data Science
Upload this notebook and `count_eligible_products.sql` to the same directory.
Use your existing `spark` session configured for OCI Object Storage. Run setup once, then the desired product cells.
Each cell executes the complete catalog selector through `spark.sql(...)` and displays its distinct eligible `NUM_IF` count.
These are source-selector counts, before destination admission and synthetic validation. Scenarios may overlap.

```python
from pathlib import Path
import re

SQL_FILE = Path("count_eligible_products.sql")  # Adjust to the uploaded file's location.
sql_text = SQL_FILE.read_text(encoding="utf-8")

count_sql = dict(re.findall(
    r"^-- BEGIN COUNT: (\w+)\s*\n(.*?)^-- END COUNT: \1\s*$",
    sql_text,
    flags=re.MULTILINE | re.DOTALL,
))
if len(count_sql) != 15:
    raise ValueError(f"Expected 15 product queries in {SQL_FILE}; found {len(count_sql)}")

# spark.sql accepts one statement per call. Remove setup comments before splitting.
setup_sql = sql_text.split("-- BEGIN COUNT:", 1)[0]
setup_sql = re.sub(r"--[^\n]*", "", setup_sql)
for statement in setup_sql.split(";"):
    if statement.strip():
        spark.sql(statement.strip())

print(f"OCI views ready. {len(count_sql)} product count queries loaded.")
```
